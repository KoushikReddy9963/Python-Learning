Input and Output

```python
import os
import cv2

#read image

image_path = os.path.join('.','data','bird.jpg')
img = cv2.imread(image_path)

#write image

cv2.imwrite(os.path.join('.','data','bird.jpg'),img)

#visualize image

cv2.imshow('image',img)
cv2.waitKey(5000) #milliseconds to show image in windows

```

```python
# similary with video
import os
import cv2

#reading a video

video_path = os.path.join('.','data','monkey,mp4')

video = cv2.VideoCapture(video_path)

#visualize video

ret = True
while ret:
    ret,frame = video.read()

    if ret:
        cv2.imshow('frame',frame)
        cv2.waitKey(40)


video.release()
cv2.destroyAllWindows()
```

```python
import cv2
#read web cam

webcam = cv2.VideoCapture(0)

#visualize webcam

while True:
    ret,frame = webcam.read()

    cv2.imshow('frame',frame)
    if cv2.waitKey(40) & 0xFF == ord('q'):
        break


webcam.release()
cv2.destroyAllWindows()
```

```python
#Basic Operations in OpenCV
#resizing
import os
import cv2

image_path = os.path.join('.','data','bird.jpg')
img = cv2.imread(image_path)

cropped_img = img[220:740,320:940]

resized_img = cv2.resize(img,(640,640)) # height width 

cv2.imshow('img',resized_img)  #here 'img is window name
cv2.imshow('cropped_img',cropped_img)
cv2.waitKey(40)

# we can also crop images 
```

Colouring images in OpenCV

```python
import os
import cv2

image_path = os.path.join('.','data','bird.jpg')
img = cv2.imread(image_path)


img_rgb = cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
img_gray = cv2.cvtColor(img,cv2.COLOR_BGR2RGB) # also HSV is there

# u can also blur the images
k_size = 7
img_blur = cv2.blur(img,(k_size,k_size))

# the more k value it more becomes blur as it takes average of cells to make blur of an image

cv2.imshow('img',img_blur)
cv2.imshow('img_rgb',img_rgb)
cv2.imshow('img_rgb',img_rgb)
```
